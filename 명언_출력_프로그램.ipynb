```python
import random

quotes = [
    "성공은 준비와 기회의 만남이다. - 세네카",
    "오늘 할 수 있는 일을 내일로 미루지 말라. - 벤자민 프랭클린",
    "작은 습관이 큰 변화를 만든다.",
    "실패는 성공으로 가는 디딤돌이다.",
    "배움은 끝없는 여행이다."
]

def show_random_quote():
    print("\n=== 오늘의 명언 ===")
    print(random.choice(quotes))

def main():
    while True:
        cmd = input("\nEnter 키를 누르면 명언을 보여줍니다 (q 입력 시 종료): ")
        if cmd.lower() == "q":
            print("프로그램을 종료합니다.")
            break
        else:
            show_random_quote()

if __name__ == "__main__":
    main()
```
